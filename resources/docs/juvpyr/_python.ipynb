```python
import importlib
import subprocess
import sys
```

```python

def packageCheck(path: str="requirements.in") -> None:
    """Installs packages from requirements.in, but avoids reinstalling if they're already installed."""

    with open(path, "r") as f:
        packages = [line.strip() for line in f.readlines() if line.strip() and not line.strip().startswith("#")]

    for package in packages:
        try:
            importlib.import_module(package)
        except ImportError:
            print(f"{package} is not installed. Installing {package}...")
            subprocess.check_call([sys.executable, "uv", "add", package])
            print(f"{package} has been installed.")
            
# %pip list
packageCheck("./requirements.txt")
```

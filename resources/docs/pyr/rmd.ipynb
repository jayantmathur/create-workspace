```r
# Cohen's effect sizes small=.02, medium=.15, large=.35 
# (Cohen, J. (1988). Statistical power analysis for the behavioral sciences (2nd ed.). Hillsdale, NJ: Lawrence Erlbaum)

# u is the number of coefficients in the model
# (i.e., number of independent + interaction variables + 1 for cutoff - b0)
# Therefore, u >= 2

# u = 3 because b0 + ConditionC + CovariateC (e.g., ManufacturabilityC)

pwr.f2.test(
    u = 2,
    f2 = cohen.ES(
        test = "t", # c("p", "t", "r", "anov", "chisq", "f2")
        size = "medium", # c("small", "medium", "large")
    )$effect.size,
    sig.level = 0.05,
    power = 0.8
)
# sample size = u + roundup(v) + 1

pwr.t.test(d = .35, sig.level = 0.05, power = 0.8, type = "one.sample")
# sample size = n


plot_theme <- plot_theme + theme(
    text = element_text(size=16),
)
```
